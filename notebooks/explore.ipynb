```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('../data/lafc_content.db')
```

```python
pd.read_sql('''
SELECT name
FROM sqlite_master
WHERE type='table';
''', conn)
```

```python
with open('../sql/videos_vs_lafc_match_context.sql') as f:
    query = f.read()
base_df = pd.read_sql(query, conn)

base_df
```

```python
import plotly.express as px

fig = px.scatter(
    base_df, x='days_since_match', y='view_count',
    log_y=True,                      # <-- fixes the skew (Problem 1)
    opacity=0.3,                     # <-- density via transparency (Problem 2)
    hover_data=['title', 'result'],  # inspect individual points
    labels={'days_since_match': 'Days since match', 'view_count': 'Views'},
)
fig.update_traces(marker=dict(size=4))
fig.update_layout(title='Views vs. days since match', template='simple_white')
fig.show()
```

```python
base_df['dur_sec'] = pd.to_timedelta(base_df['duration']).dt.total_seconds()
fig = px.scatter(base_df, x='dur_sec', y='view_count', log_y=True, opacity=0.3, hover_data='title')
fig.show()
```
